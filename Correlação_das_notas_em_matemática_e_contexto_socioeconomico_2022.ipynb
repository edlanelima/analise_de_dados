```python
import numpy as np
import statsmodels.api as sm
from scipy import stats

X = df22_publica_sudeste[colunas_interesse]
y = df22_publica_sudeste['NU_NOTA_MT']

# Adicione uma coluna de uns para a constante na regressão
X = sm.add_constant(X)

# Ajuste o modelo de regressão
modelo = sm.OLS(y, X).fit()

# Calcule os resíduos do modelo
residuos = modelo.resid

# Realize o teste de Shapiro-Wilk nos resíduos
statistic, p_value = stats.shapiro(residuos)

# Imprima os resultados
print("Estatística de Teste:", statistic)
print("Valor-p:", p_value)

# Faça a interpretação
alpha = 0.05
if p_value > alpha:
    print("Não há evidências para rejeitar a hipótese nula (os resíduos parecem seguir uma distribuição normal)")
else:
    print("As evidências sugerem que os resíduos não seguem uma distribuição normal")

```
